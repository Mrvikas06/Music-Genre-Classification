## 01. Initiate Exploratory Data Analysis by Loading Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set overall plot style
sns.set_theme(style="whitegrid")

# Load 30-second feature dataset
df = pd.read_csv('../data/features_30_sec.csv')

# Display first 5 rows
df.head()
```

## 02. Perform Initial DataFrame Inspection with `.info()` and `.describe()`

```python
# Summary of columns, non-null counts, and data types
df.info()
```

```python
# Statistical summary for numerical features
df.describe()
```

## 03. Check for Missing Values in a Pandas DataFrame

```python
# Check for missing / NaN values across all columns
missing_values = df.isnull().sum()
print("Missing values per column:")
if (missing_values > 0).any():
    print(missing_values[missing_values > 0])
else:
    print("No missing values found in the dataset! Total nulls: 0")
```

## 04. Visualize Music Genre Distribution with a Bar Plot

```python
# Plot genre distribution to verify class balance
plt.figure(figsize=(10, 4))
sns.countplot(data=df, x='label', palette='viridis')
plt.title('Music Genre Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Genre', fontsize=12)
plt.ylabel('Track Count', fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 05. Visualize Feature Distribution with a Box Plot

```python
# Box plot showing tempo variation across genres
plt.figure(figsize=(12, 5))
sns.boxplot(data=df, x='label', y='tempo', palette='coolwarm')
plt.title('Tempo Distribution Across Music Genres', fontsize=14, fontweight='bold')
plt.xlabel('Genre', fontsize=12)
plt.ylabel('Tempo (BPM)', fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 06. Generate Violin Plot for First MFCC Across Music Genres

```python
# Violin plot showing probability density and distribution of mfcc1_mean across genres
plt.figure(figsize=(12, 5))
sns.violinplot(data=df, x='label', y='mfcc1_mean', palette='muted')
plt.title('MFCC 1 Distribution Across Music Genres', fontsize=14, fontweight='bold')
plt.xlabel('Genre', fontsize=12)
plt.ylabel('MFCC 1 Mean', fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 07. Compute the Correlation Matrix

```python
# Select numeric columns only (excluding 'filename' and 'label')
numeric_df = df.select_dtypes(include=[np.number])

# Compute Pearson correlation matrix
corr_matrix = numeric_df.corr()
corr_matrix.head(10)
```

## 08. Visualize Feature Correlations with a Heatmap

```python
# Select key representative features for an interpretable heatmap
selected_features = [
    'chroma_stft_mean', 'rms_mean', 'spectral_centroid_mean',
    'spectral_bandwidth_mean', 'rolloff_mean', 'zero_crossing_rate_mean',
    'harmony_mean', 'perceptr_mean', 'tempo', 'mfcc1_mean', 'mfcc2_mean'
]

plt.figure(figsize=(10, 8))
sns.heatmap(
    df[selected_features].corr(),
    annot=True,
    fmt=".2f",
    cmap='coolwarm',
    linewidths=0.5
)
plt.title('Correlation Heatmap of Audio Features', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
